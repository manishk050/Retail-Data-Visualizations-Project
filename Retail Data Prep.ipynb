# Retail Sales Data Preparation and Validation

This notebook prepares the 2024, 2025 and 2026 retail transaction files for analysis. It cleans and combines the annual data, adds product and customer details from the master tables, checks data quality, and exports an integrated dataset for Power BI.

## 1. Load and inspect the annual transaction data


Import the required libraries and annual order files, then inspect and prepare the transaction data before combining the three years.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
retail_2024 = pd.read_excel("BITS_Retail_2024.xlsx")
retail_2025 = pd.read_excel("BITS_Retail_2025.xlsx")
retail_2026 = pd.read_excel("BITS_Retail_2026.xlsx")
```

### Inspect data types and completeness


Review the structure of each annual file, including column types and the number of non-null values. This helps identify fields that need attention before combining the data.

```python
retail_2024.info()
```

```python
retail_2025.info()
```

```python
retail_2026.info()
```

### Set appropriate data types


Convert identifiers to text, dates to datetime, and numeric fields to suitable numeric types so that the files can be combined and analyzed consistently.

```python
datasets = [retail_2024, retail_2025, retail_2026]

for df in datasets:
    df["Order_ID"] = df["Order_ID"].astype("string")
    df["Customer_ID"] = df["Customer_ID"].astype("string")
    df["Product_ID"] = df["Product_ID"].astype("string")
    df["Region"] = df["Region"].astype("category")
    
    df["Order_Date"] = pd.to_datetime(df["Order_Date"])
    df["Quantity"] = df["Quantity"].astype("int64")
    df["Unit_Price"] = df["Unit_Price"].astype("float64")
    df["Discount_Pct"] = df["Discount_Pct"].astype("float64")
    df["Shipping_Cost"] = df["Shipping_Cost"].astype("float64")
```

### Check duplicate rows and missing values


Profile each annual file for exact duplicate rows and missing values before removing or changing any records.

```python
for year, df in zip([2024, 2025, 2026], datasets):

    print(f"\n--- {year} ---")
    print("Original rows:", len(df))
    print("Duplicate rows:", df.duplicated().sum())
    print("Missing values:\n", df.isna().sum())
    print("Rows containing missing values:",
          df.isna().any(axis=1).sum())
```

### Remove duplicate and incomplete records


Remove exact duplicate rows and rows containing missing values, then review the cleaned annual files before appending them.

```python
for df in datasets:
    df.drop_duplicates(inplace=True)
    df.dropna(inplace=True)
```

### Review the cleaned annual files


Preview the cleaned 2024, 2025 and 2026 tables to confirm that the records and columns look as expected before appending.

```python
retail_2024.head()
```

```python
retail_2025.head()
```

```python
retail_2026.head()
```

### Append the annual transaction files


Combine the cleaned 2024, 2025 and 2026 transaction tables into one dataset with a consistent column structure.

```python
retail_sales = pd.concat(datasets, ignore_index=True)

retail_sales.head()
```

```python
retail_sales.shape
```

### Check order dates against the source year


Verify that each transaction's order date belongs to the year of its source file. Any exceptions should be investigated rather than changed automatically.

```python
for year, df in zip([2024, 2025, 2026], datasets):
    mismatched_dates = (df["Order_Date"].dt.year != year).sum()
    print(f"{year} - dates outside expected year: {mismatched_dates}")
```

## 2. Load and prepare the master data


Inspect the product and customer lookup tables, align their data types with the transaction data, and standardize known category labels.

### Load the product and customer master tables


Import the master workbook and separate the product and customer lookup tables that will be used to enrich the transaction data.

```python
master_sheets = pd.read_excel("BITS_Retail_Master.xlsx", sheet_name=None)
master_product = master_sheets["Product_Master"]
master_customer = master_sheets["Customer_Master"]
```

```python
master_product.head()
```

```python
master_customer.head()
```

### Inspect the master tables


Review the master tables' columns, data types and completeness before using them in the joins.

```python
master_product.info(),master_customer.info()
```

```python
print("\nCategory value counts:")
print(master_product["Category"].value_counts(dropna=False))
```

### Align data types and standardize category labels


Make the master-table keys compatible with the transaction data and correct the known category-label variations before merging.

```python
# Align master table data types with the retail sales data
master_product["Product_ID"] = master_product["Product_ID"].astype("string")
master_product["Product_Name"] = master_product["Product_Name"].astype("string")

# Standardize known category-label inconsistencies before merging
category_map = {
    "Tech": "Technology",
    "Office Suply": "Office Supplies",
    "Office Suppllies": "Office Supplies"
}

master_product["Category"] = (
    master_product["Category"]
    .astype("string")
    .replace(category_map)
    .astype("category")
)

master_product["Subcategory"] = master_product["Subcategory"].astype("category")
master_product["Unit_Cost"] = master_product["Unit_Cost"].astype("float64")

master_customer["Customer_ID"] = master_customer["Customer_ID"].astype("string")
master_customer["Customer_Segment"] = master_customer["Customer_Segment"].astype("category")

master_product.info()
master_customer.info()
```

```python
master_product["Category"].value_counts()
```

## 3. Integrate the transaction and master data


Enrich each transaction with product and customer attributes, then verify the quality and completeness of the joins.

### Enrich transactions with product and customer details


Join the product and customer attributes to each transaction using their respective IDs. The many-to-one validation helps confirm that each master key maps to no more than one master record.

```python
retail_sales_enriched = (
    retail_sales
    .merge(master_product, on="Product_ID", how="left", validate="many_to_one")
    .merge(master_customer, on="Customer_ID", how="left", validate="many_to_one")
)

retail_sales_enriched.head()
```

### Check product and customer key matches


Identify transaction records whose product or customer IDs do not appear in the corresponding master table. These unmatched keys need to be reviewed.

```python
# Check unmatched product IDs
unmatched_products = retail_sales[
    ~retail_sales["Product_ID"].isin(
        master_product["Product_ID"]
    )
]

# Check unmatched customer IDs
unmatched_customers = retail_sales[
    ~retail_sales["Customer_ID"].isin(
        master_customer["Customer_ID"]
    )
]

print("Unmatched products:", len(unmatched_products))
print("Unmatched customers:", len(unmatched_customers))
```

### Verify row counts after merging


Compare the transaction count before and after the master-table joins to confirm that the merges have not unexpectedly added or removed transaction rows.

```python
print("Before merge:", len(retail_sales))
print("After merge:", len(retail_sales_enriched))
```

### Check missing values after the merge


Check the enriched dataset for missing values, which may indicate incomplete source data or master records that did not match.

```python
missing_after_merge = retail_sales_enriched.isna().sum()

print("Total missing values after merge:", missing_after_merge.sum())
print("Columns with missing values:")
print(missing_after_merge[missing_after_merge > 0])
```

## 4. Validate the integrated dataset


Run final checks on numerical fields, categorical values, data types and duplicates before creating the export file.

### Review numerical field distributions


Inspect summary statistics for quantity, price, discount, shipping cost and unit cost to identify values that may need further investigation.

```python
numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount_Pct",
    "Shipping_Cost",
    "Unit_Cost"
]

print(retail_sales_enriched[numeric_columns].describe())
```

### Check numeric fields against basic business rules


Flag non-positive quantities, prices or costs, discounts outside the expected range, and negative shipping costs. These checks identify values for review; they do not automatically remove records.

```python
df = retail_sales_enriched

numeric_checks = {
    "Non-positive quantities": (df["Quantity"] <= 0).sum(),
    "Non-positive unit prices": (df["Unit_Price"] <= 0).sum(),
    "Non-positive unit costs": (df["Unit_Cost"] <= 0).sum(),
    "Discounts outside 0 to 1": ((df["Discount_Pct"] < 0) | (df["Discount_Pct"] > 1)).sum(),
    "Negative shipping costs": (df["Shipping_Cost"] < 0).sum()
}

for check, count in numeric_checks.items():
    print(f"{check}: {count}")
```

```python
retail_sales_enriched.info()
```

### Review categorical values


Inspect the distinct values and record counts for region, category, subcategory and customer segment to confirm that the labels are consistent.

```python
print("Region value counts:")
print(retail_sales_enriched["Region"].value_counts(dropna=False))
print("\nCategory value counts:")
print(retail_sales_enriched["Category"].value_counts(dropna=False))
```

```python
print("Subcategory value counts:")
print(retail_sales_enriched["Subcategory"].value_counts(dropna=False))
print("\nCustomer Segment value counts:")
print(retail_sales_enriched["Customer_Segment"].value_counts(dropna=False))
```

### Apply consistent precision to monetary fields


Round the selected monetary fields to two decimal places for consistent presentation. The discount percentage is retained at its original precision because it is a proportion, not a currency amount.

```python
# Keep monetary fields to two decimal places.
# Discount_Pct is a proportion, so it is not included in this rounding step.
currency_columns = [
    "Unit_Price",
    "Shipping_Cost",
    "Unit_Cost",
]

for col in currency_columns:
    if col in retail_sales_enriched.columns:
        retail_sales_enriched[col] = retail_sales_enriched[col].round(2)
```

### Run final duplicate checks


Check the integrated dataset for duplicate rows and repeated order IDs before exporting the final file.

```python
duplicate_row_count = retail_sales_enriched.duplicated().sum()
duplicate_order_id_count = retail_sales_enriched["Order_ID"].duplicated().sum()

print(f"Duplicate rows: {duplicate_row_count}")
print(f"Duplicate Order_ID values: {duplicate_order_id_count}")
```

Save the integrated transaction-level data as a CSV file so it can be imported into Power BI for measure creation and visualization.
